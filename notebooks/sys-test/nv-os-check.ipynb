# NVIDIA system check

```python
%%bash

(type nvidia-smi && nvidia-smi -L) &> /dev/null \
    && X_HAS_GPU=1 || X_HAS_GPU=0;

case "$X_HAS_GPU" in
     1) export X_UV_EXTRA='gpu' ;;
     *) export X_UV_EXTRA='cpu' ;;
esac
export X_HAS_GPU

echo "X_HAS_GPU=$X_HAS_GPU"
echo "X_UV_EXTRA=$X_UV_EXTRA"
```
