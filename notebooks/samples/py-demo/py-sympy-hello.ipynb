```python
from sympy import init_printing, symbols, Integral, integrate, oo, exp

init_printing()
```

```python
x, y, z, t = symbols("x y z t")
```

```python
f = exp(-(x**2) - y**2)
F = f , (x, -oo, oo), (y, -oo, oo)
```

```python
Integral(*F)
```

```python
integrate(*F)
```
